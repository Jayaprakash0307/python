Searching Algorithms


1)Linear Search

```python
def linear_search(lst,target):
    size = len(lst)
    for index in range(size):
        if lst[index] == target:
            return index
    return -1

lst = [10,20,30,40,50]
target =10
result = linear_search(lst,target)
print(result)
```
