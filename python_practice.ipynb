Check  whether a given year is leap year or not

```python
year = int(input())
if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("the given year is a leap year ")
        else:
            print("the given year is not a leap year")
    else:
        print("the given year is a leap year")
else:
    print("the given year is not a leap year")
```

```python
num = int(input())
original_num = num
rev = 0
while num > 0:
    digit = num % 10
    rev = rev*10 + digit
    num = num//10
if rev == original_num:
    print("the given number is palindrome")
else:
    print("the given number is not a palindrome")
```
